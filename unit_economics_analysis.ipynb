### Финансовые метрики (Financial Metrics)

**1. Выручка (Revenue)**
$$Revenue = price \times (1 - discount) \times quantity$$
* Рассчитывается с учетом скидки на единицу товара.

**2. Себестоимость продаж (COGS)**
$$COGS = cost \times quantity$$
* Общие затраты на производство или закупку проданного объема.

**3. Валовая прибыль (Gross Profit)**
$$Gross Profit = Revenue - COGS$$
* Абсолютный доход после вычета прямых производственных затрат.

**4. Валовая маржинальность (Gross Margin, %)**
$$Gross Margin = \frac{Gross Profit}{Revenue} \times 100\%$$
* Доля валовой прибыли в общем объеме выручки.


````python
sql_query = """
SELECT
  category,
  SUM(price * (1 - discount) * quantity) AS Revenue,
  SUM(cost * quantity) AS COGS,
  SUM(price * (1 - discount) * quantity - (cost * quantity)) AS 'Gross Profit',
  SUM(price * (1 - discount) * quantity - (cost * quantity)) / SUM (price * (1 - discount) * quantity)* 100 AS 'Gross Margin'
FROM orders_merged
GROUP BY category;
"""
df = pd.read_sql_query(sql_query, conn)
df.head(10)
````


# Ошибки

## Ошибка №1

### Доля от общего (SQL)

Знаменатель для **доли от общего** — это независимый подзапрос без `GROUP BY`.

**Формула расчета:**
```sql
SUM(формула) / (SELECT SUM(формула) FROM таблица) * 100
```

* **Числитель** живёт внутри текущей группировки.
* **Знаменатель** полностью игнорирует группировку.
* Деление происходит между **двумя завершёнными агрегатами**, а не внутри одного `SUM`.

### Логика выполнения скобок в SQL

**Скобки** = *"выполни это первым и получи одно число"*.

* Всё, что стоит **вне скобок** и **после** закрывающей `)`, применяется к уже готовому результату.
* Внешние операции **никогда** не применяются к тому, что находится внутри скобок в процессе их расчета.



## Ошибка №2

### Логика расчета Profit Contribution (Аналог Задачи 4)

Вам нужна **независимая величина**, которая сама по себе требует предварительной группировки внутри, а уже затем — усреднения.

❌ **Почему НЕ работает одна строка:**
```sql
AVG(price * (1 - discount) * quantity)
```
Эта функция считает среднее по каждой отдельной строке (транзакции), а не по сгруппированному товару.

**Правильный подход:**
Нужен **отдельный подзапрос**, который:
1. Сначала внутри себя **группирует по товару** и **суммирует** выручку.
2. Затем снаружи берет **`AVG()`** от готовых результатов этой группировки.


**Пример SQL-запроса с вложенной агрегацией:**

```sql
(
  SELECT
    AVG(revenue_товара)
  FROM (
    SELECT
      SUM(price * (1 - discount) * quantity) AS revenue_товара
    FROM orders_merged
    GROUP BY product_name
  )
)
```

### ⚠️ Важный нюанс: Строка ≠ Заказ (Переход к AOV)

**Среднее по строкам — это не то же самое, что средняя выручка на заказ!**  
Если в одном заказе куплено несколько товаров (несколько строк с одинаковым `order_id`), то расчет по строкам исказит метрику **AOV** (Average Order Value).

В аналитике данных мы разделяем **три разные единицы измерения**:

1. 📄 **Строка** — одна конкретная позиция товара в чеке / заказе.
2. 📦 **Товар** — уникальный артикул (`product_name`), суммарно по всем его продажам.
3. 🛒 **Заказ** — корзина покупателя (`order_id`), суммарно по всем позициям внутри неё.


### 🔑Двойная агрегация через `FROM (SELECT ...)`

Конструкция `FROM (SELECT ...)` нужна всегда, когда требуется применить агрегатную функцию (`SUM`, `AVG`, `COUNT`, `MAX`, `MIN`) к результату, который **уже сам является агрегатом**.

> ⚠️ **Главное ограничение:** Одна агрегатная функция не может стоять внутри другой в рамках одного `SELECT`. Поэтому операции физически разносят на два последовательных шага.

---

#### 📌 Паттерн 1: "Среднее/сумма от агрегатов по группам"
* **Когда применяется:** Нужна метрика уровня «выше», чем группа, которую вы уже посчитали (например, «среднее по товарам», когда товар сам — результат `SUM` по строкам).
* **Пример:** Задача 6 (средняя выручка товара, а не строки).

```sql
SELECT
  AVG(revenue)
FROM (
  SELECT
    SUM(...) AS revenue
  FROM orders_merged
  GROUP BY product_name
)
```

---

#### 📌 Паттерн 2: "Посчитать метрику на одном уровне, потом фильтровать/агрегировать"
* **Когда применяется:** Нужно сначала построить промежуточную таблицу с метриками (например, `gross_margin` по товару), а затем выполнить с ней операции — отфильтровать, отсортировать или агрегировать дальше.
* **Пример:** «Чистая» альтернатива для Задачи 5. Вы считаете всё во внутреннем `SELECT` с понятными алиасами, а во внешнем — просто фильтруете по готовым колонкам.

```sql
SELECT
  category,
  gross_margin
FROM (
  SELECT
    category,
    SUM(...) / SUM(...) * 100 AS gross_margin
  FROM orders_merged
  GROUP BY category
)
WHERE gross_margin > 50
```
*(Здесь во внешнем запросе уже **можно использовать алиасы**, так как внутренний `SELECT` — это завершенный, самостоятельный контекст).*

---

#### 📌 Паттерн 3: "Сначала детальный расчет, затем крупная группировка"
* **Когда применяется:** Нужна метрика на детальном уровне (заказа/товара) как промежуточный шаг, а финальная группировка идет по более крупной сущности (категория, клиент).
* **Пример:** «Средний чек по категориям», если в одном заказе могут быть товары разных категорий.

```sql
SELECT
  category,
  AVG(order_total)
FROM (
  SELECT
    order_id,
    category,
    SUM(...) AS order_total
  FROM orders_merged
  GROUP BY order_id, category
)
GROUP BY category
```

---

#### 💡 Ключевой вопрос для самопроверки перед написанием SQL:

> *"Единица, которую я хочу усреднить/просуммировать — это уже результат агрегации или сырая строка?"*

* **Сырая строка** ➡️ достаточно одного `SUM`/`AVG` с обычным `GROUP BY`.
* **Уже агрегат** (сумма/среднее чего-то, что считалось через `GROUP BY`) ➡️ строго нужен `FROM (SELECT ... GROUP BY ...)`, то есть двойной уровень.


##Пояснение к задаче 11: Когда вложенность избыточна, а когда обязательна

Перед написанием запроса проверьте физическое совпадение единицы анализа и строки таблицы с помощью:
```sql
SELECT COUNT(*) FROM таблица GROUP BY <единица>
```

---

#### 🟢 Сценарий 1: Единица физически совпадает со строкой
*(Если результат проверки `COUNT(*)` везде равен **1**)*

* **Суть:** В каждой строке таблицы лежит ровно один уникальный объект (заказ/товар/клиент).
* **Решение:** Двухуровневая вложенность `FROM (SELECT ...)` даст правильный, но **избыточный результат**.
* **Как писать:** Можно и нужно писать проще — одним `AVG(...)` или `SUM(...)` с обычным `GROUP BY`, без использования вложенных подзапросов.

---

#### 🔴 Сценарий 2: Единица НЕ совпадает со строкой
*(Если результат проверки `COUNT(*)` выдает значения **больше 1**)*

* **Суть:** У одной единицы анализа есть несколько строк в таблице (например, несколько позиций товаров внутри одного `order_id`).
* **Решение:** Вложенность `FROM (SELECT ... GROUP BY <единица>)` **строго обязательна**.
* **Результат без вложенности:** Если посчитать обычный `AVG(...)` по сырым строкам, результат будет гарантированно **неверным**, так как крупные заказы/популярные товары исказят среднее значение (как в Задаче 6 со средней выручкой товара).
